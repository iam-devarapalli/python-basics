# Python Setup in VS Code

## 1. Install the tools
- Install Python: https://www.python.org/downloads/
- Install VS Code: https://code.visualstudio.com/
- Install the Python and Jupyter extensions by Microsoft.

## 2. Create a virtual environment

Run these commands in the VS Code terminal.

### Mac
```bash
python3 -m venv .venv
source .venv/bin/activate
```

### Windows — PowerShell
```powershell
py -m venv .venv
.\.venv\Scripts\Activate.ps1
```

## 3. Install notebook support
With the virtual environment active, run:

```bash
python -m pip install ipykernel
```

## 4. Select the notebook kernel
Click Select Kernel → Python Environments → choose .venv.

Check which Python your notebook uses:

````python
import sys

print("Python version:", sys.version)
print("Python location:", sys.executable)
````

| Operating system | terminal | Activation command |
